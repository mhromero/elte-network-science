```python
import requests, gzip
from io import BytesIO

# URL for the Facebook Ego Network dataset
url = "https://snap.stanford.edu/data/facebook_combined.txt.gz"

# Download and decompress
response = requests.get(url)
file_like = BytesIO(response.content)
data = gzip.decompress(file_like.read())

# Create file-like object
file_like = BytesIO(data)
```

```python
import networkx as nx
path = "./facebook_combined_with_metrics.graphml"
G = nx.read_graphml(path)
```

```python
n = 5
top_nodes_dc = sorted(metrics["degree_centrality"].items(), key=lambda x: x[1], reverse=True)[:5]
G_copy = G.copy()
G_copy.remove_nodes_from(top_nodes_dc)
LCC_after_attack = max(nx.connected_components(G), key=len)
LCC_after_attack
```

```python
def simulate_attack(G, metric, n):
    top_nodes_dc = [node for node, _ in sorted(metric.items(), key=lambda x: x[1], reverse=True)[:n]]
    G_copy = G.copy()
    G_copy.remove_nodes_from(top_nodes_dc)
    LCC_after_attack = max(nx.connected_components(G_copy), key=len)
    LCC_after_attack

    original_LCC = max(nx.connected_components(G), key=len)
    # now we compare the subgraphs
    original_LCC_subgraph = G.subgraph(original_LCC).copy()
    attacked_LCC_subgraph = G_copy.subgraph(LCC_after_attack).copy()
    
    result = {
        "original_LCC_size": original_LCC_subgraph.number_of_nodes(),
        "attacked_LCC_size": attacked_LCC_subgraph.number_of_nodes(),
        "original_LCC_edges": original_LCC_subgraph.number_of_edges(),
        "attacked_LCC_edges": attacked_LCC_subgraph.number_of_edges(),
        "node_loss_frac": (original_LCC_subgraph.number_of_nodes() - attacked_LCC_subgraph.number_of_nodes())/original_LCC_subgraph.number_of_nodes(),
        "edge_loss_frac": (original_LCC_subgraph.number_of_edges() - attacked_LCC_subgraph.number_of_edges())/original_LCC_subgraph.number_of_edges()
    }
    return result
```

```python
n = 15
print("Attack simulations")
print("Degree Centrality Attack:")
print(simulate_attack(G, metrics["degree_centrality"], 15))

print("Betweenness Centrality Attack:")
simulate_attack(G, metrics["betweenness_centrality"], 15)
```
