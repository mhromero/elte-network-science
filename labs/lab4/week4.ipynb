```python
nodes = [("Node A", "Start"), ("Node B", "Process"), ("Node C", "End")]
edges = [("Node A", "Node B"), ("Node B", "Node C")]

graph = graphviz.Digraph()

for name, label in nodes:
    graph.node(name, label)

graph.edges(edges)

graph.render("graph1", format="png", view=True)

```

```python
print(graph.source)
```

```python
graph.node?
```

```python
graph.node("Node A", "Start", fillcolor="green", style="filled", shape="circle")
graph.node("Node B", "Process", color="black", fillcolor="lightblue", style="filled", shape="box")
graph.node("Node C", "End", shape="diamond")

graph.render("graph2", format="png", view=True)
```

```python
# try out different render engines
engines = ["dot", "neato", "fdp", "sfdp", "twopi", "circo"]
for engine in engines:
    graph.render(f"graph_{engine}", format="png", engine=engine, view=True)
```

```python
graph.edge("Node A", "Node C")
graph.render("graph_extra_edge", format="png", view=True)
```

```python
nolabel_graph = graphviz.Digraph()
for name, label in nodes:
    nolabel_graph.node(name)

nolabel_graph.edges(edges)

nolabel_graph.render("unlabeled_graph", format="png", view=True)
```

```python
print(nolabel_graph.source)
```

```python
import graphviz
expanded_graph = graphviz.Digraph()

nodes = [("Node 1", "1"), ("Node 2", "2"), ("Node 3", "3"), ("Node 4", "4"), ("Node 5", "5")]
edges = [("Node 1", "Node 2"), ("Node 2", "Node 3"), ("Node 3", "Node 4"), ("Node 4", "Node 5")]

for name, label in nodes:
    expanded_graph.node(name, label, fontsize="20")
expanded_graph.edges(edges)

expanded_graph.render("expanded_graph_dot", format="png", view=True, engine="dot")
expanded_graph.render("expanded_graph_circo", format="png", view=True, engine="circo")
```

```python
expanded_graph.node("Node 1", "1", fillcolor="pink", style="filled", shape="doublecircle", fontsize="20")
expanded_graph.edge("Node 1", "Node 5", label="extra", style="dashed")
expanded_graph.render("expanded_graph_circo", format="png", view=True, engine="circo")
```

```python
workflow = graphviz.Digraph(name="Data Pipeline")

nodes = [("1", "Collect Data"), ("2", "Clean Data"), ("3", "Train Model"), ("4", "Evaluate"), ("5", "Deploy")]
edges = [("1", "2"), ("2", "3"), ("3", "4"), ("4", "5"), ("4", "2")]

for name, label in nodes:
    workflow.node(name, label)

workflow.edges(edges)
workflow.node("3", "Train Model", shape="box", style="filled", fillcolor="yellow", color="black")

workflow.render("data_pipeline", format="png", view=True, engine="dot")
workflow.render("data_pipeline_circo", format="png", view=True, engine="circo")
```
